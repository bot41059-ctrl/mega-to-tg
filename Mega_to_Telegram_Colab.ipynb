- 🔍 **Deep Scanner:** Sabhi folders aur subfolders ka hierarchy tree scan karta hai.

### ⚙️ Step 2: Bot Configuration
Niche diye gaye form me apni Telegram API details aur Target Channel ID bhariye:
1. **`BOT_TOKEN`**: [@BotFather](https://t.me/BotFather) se mila hua token.
2. **`API_ID` & `API_HASH`**: [my.telegram.org](https://my.telegram.org) se mila hua API ID aur Hash.
3. **`CHANNEL_ID`**: Wo Telegram Channel jaha videos upload karni hai (e.g. `-1001234567890` ya `@MyChannel`). *(Make sure bot channel me Admin ho!)*

```python
#@title ⚙️ Run Step 2: Configure Bot Credentials
BOT_TOKEN = "" #@param {type:"string"}
API_ID = 0 #@param {type:"integer"}
API_HASH = "" #@param {type:"string"}
CHANNEL_ID = "" #@param {type:"string"}
AUTH_USERS = "*" #@param {type:"string"}

assert BOT_TOKEN, "⚠️ Please enter your BOT_TOKEN from @BotFather!"
assert API_ID != 0, "⚠️ Please enter your API_ID from my.telegram.org!"
assert API_HASH, "⚠️ Please enter your API_HASH from my.telegram.org!"

# Write configuration directly into .env file
env_content = f"""# Generated Bot Configuration
BOT_TOKEN={BOT_TOKEN.strip()}
APP_ID={API_ID}
API_ID={API_ID}
API_HASH={API_HASH.strip()}
CHANNEL_ID={CHANNEL_ID.strip()}
AUTH_USERS={AUTH_USERS.strip()}
DOWNLOAD_LOCATION=/content/downloads
TG_MAX_SIZE=2040108421
SPLITTER_BUFFER_SIZE=1048576
"""

with open(".env", "w", encoding="utf-8") as f:
    f.write(env_content.strip() + "\n")

print("✅ Configuration saved to .env file!")
if CHANNEL_ID:
    print(f"📢 Target Channel set to: {CHANNEL_ID}")
else:
    print("ℹ️ No channel configured yet. You can set it in Telegram anytime using /setchannel command.")
```
